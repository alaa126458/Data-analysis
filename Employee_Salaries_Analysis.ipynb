```python
import pandas as pd
```

```python
df=pd.read_csv("Employee_Salaries.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
#What is the average base salary?
df['Base_Salary'].mean()
```

```python
#what is the highest amount of overtime pay?
df['Overtime_Pay'].max()
```

```python
df['BasePay_benefit']=df['Base_Salary']+df['Overtime_Pay']
df
```

```python
#what is the name of department name make the highest (benefits)?
df[df['BasePay_benefit']==df['BasePay_benefit'].max()]['Department_Name']
```

```python
#what is name of department name make the lowest(benefits)?
df[df['BasePay_benefit']==df['BasePay_benefit'].min()]['Department_Name']
```

```python
#what is the average base pay of all department name for all gender?
df.groupby('Gender')['Base_Salary'].mean()
```

```python
#how many department name are there?
df['Department_Name'].nunique()
```

```python
#what are the top 5 Common department name?
df['Department_Name'].value_counts().head()
```
